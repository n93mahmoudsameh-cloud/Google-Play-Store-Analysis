Step 4 : Data Cleaning

```python
# Remove Duplicates
Google_Play.drop_duplicates(inplace=True)
```

```python
print('Total Duplicats :\n' , Google_Play.duplicated().sum())
```

```python
# Remove Null Values
Google_Play.dropna(inplace=True)
```

```python
print('Total Null Values :\n',Google_Play.isnull().sum().sum())
```

```python
# Change Data Type

Google_Play['Price'] = Google_Play['Price'].str.replace('$', '', regex=False)
Google_Play['Price'] = Google_Play['Price'].astype(float)

Google_Play['Last Updated'] = pd.to_datetime(Google_Play['Last Updated'])
```

```python
# Handling Columns

Google_Play['Installs'] = (
    Google_Play['Installs']
    .str.replace('+', '', regex=False)
    .str.replace(',', '', regex=False)
    .astype(int)
)
```

Step 5 : Feature Engineering

```python
# Export Totla Sales

Google_Play['Total Sales']=sum(Google_Play['Price'])
```
